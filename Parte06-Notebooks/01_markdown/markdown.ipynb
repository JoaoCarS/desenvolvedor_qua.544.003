# Título do meu documento

Nesta aula iremos aprender sobre como inserir ***markdowns*** dentro de arquivos do tipo **Jupyter Notebook**.

> ***Markdown*** é um tipo de documento muito usado em ciência de dados e também para alimentar Inteligencia Artificial. 
>
>Manipular este tipo de arquivo se torna fundamental para dominar as profissões do futuro que dependem de IAs.

## Subtítulo de markdown

O que diferencia um markdown de um notebook, é que no caso do notebook, ele também é capaz de ler códigos das linguagens de programação *R*,*Julia*, e é claro, *Python*.
